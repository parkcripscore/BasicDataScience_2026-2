- lambda를 이용하여 데이터 프레임 이름 줄여 쓰기
  - assign( ) 함수를 사용할 때 데이터 프레임 이름을 반복해서 입력해야 하므로 번거로움
  - 이름이 긴 경우 그 이름을 일일이 입력하는 것이 불편할 수 있음
  - lambda를 이용하면 약어를 이용하여 코드를 간결하게 작성할 수 있음

```python
long_name = pd.read_csv('exam.csv')   # 위와 같은 파일

# long_name 직접 입력
l1 = long_name.assign(new = long_name['math'] + long_name['english'] + long_name['science'])
print(l1)

print("==================================================")

# long_name 대신 x 입력
l2 = long_name.assign(new = lambda x: x['math'] + x['english'] + x['science'])
print(l2)
```

- 여러 개의 파생 변수를 만들 때 앞에서 만든 변수를 활용해 다시 변수 만들기

### 집단별로 요약하기
- 집단 별 평균이나 집단 별 빈도처럼 각 집단을 요약한 값을 구하는 경우 groupby( ) 함수와 agg( ) 함수 사용
- 집단 간에 어떤 차이가 있는지 쉽게 파악 가능

- 전체 요약 통계량 구하기

```python
import pandas as pd

exam = pd.read_csv('exam.csv')

# 파일이 없다면 아래 두 줄의 주석을 해제하여 사용
# url = 'https://raw.githubusercontent.com/skang-koreatech/BasicDataScience_2026-2/main/exam.csv'
# exam = pd.read_csv(url)

# agg( ) 함수의 입력 : 요약값을 할당할 변수명 = ('사용할 변수명', '함수명')
agg1 = exam.agg(mean_math = ('math', 'mean'))
print(agg1)
```

- 집단 별 요약 통계량 구하기
  - groupby( ) 함수 : 데이터를 범주로 나누어 구분 
  - agg( ) 함수 : 구분된 그룹별로 통계량 계산

### pandas 함수 조합하기 (메소드 체이닝, method chaining)

- 대부분의 pandas 함수는 원본을 바꾸지 않고 **처리 결과를 돌려줌**
  - 따라서 그 결과에 다시 점을 찍어 다음 함수를 이어 붙일 수 있음
  - 이렇게 여러 함수를 연달아 연결하는 방식을 **메소드 체이닝**이라 함

- 돌려주는 것이 항상 DataFrame인 것은 아님
  - 예를 들어 `groupby( )` 는 그룹 객체를, `df['math']` 는 Series를 돌려줌
  - 무엇이 돌아오든 그 다음에 이어 붙일 수 있는 함수가 정해져 있을 뿐, 체이닝은 그대로 성립함

- 중간 결과를 변수에 담아 가며 단계별로 작성해도 결과는 같음

| 단계별로 작성 | 체이닝으로 작성 |
|---|---|
| `a = mpg.query(...)` | `mpg.query(...)` |
| `b = a.assign(...)` | `    .assign(...)` |
| `c = b.groupby(...)` | `    .groupby(...)` |
| `print(c)` | |

- 체이닝의 장점
  - 이름만 붙이고 한 번만 쓰는 중간 변수(`a`, `b`, `c`)를 만들지 않아도 됨
  - 처리 **순서가 위에서 아래로 그대로 읽힘**

- 주의할 점
  - 줄을 나누어 쓰려면 줄 끝에 `\` 를 붙여 아직 문장이 끝나지 않았음을 표시해야 함
  - 체이닝 중간에서는 **원본 데이터의 이름을 다시 쓰면 안 됨**
    - 앞 단계를 거치며 행 수가 달라져 있기 때문
    - 앞 단계까지의 결과를 가리키려면 `lambda x` 의 `x` 를 사용

##### Q. mpg 데이터에서 제조 회사별로 'suv' 자동차의 도시 및 고속도로 합산 연비 평균을 구해 내림차순으로 정렬하고 1~5위까지 출력하기

- 필요한 절차
  - 'suv' 카테고리 자동차 추출하기
  - 합산 연비 변수 만들기 (`cty` + `hwy`)
  - 제조 회사별로 구분하기
  - 합산 연비 평균 구하기
  - 내림차순 정렬하기
  - 1~5위까지 출력하기

```python
# 단계별로 작성한 코드

# 1. 'suv' 카테고리 자동차 추출하기
step1 = mpg.query('category == "suv"')

# 2. 합산 연비 변수 만들기
step2 = step1.assign(total = step1['cty'] + step1['hwy'])

# 3. 제조 회사별로 구분하기
step3 = step2.groupby('manufacturer')

# 4. 합산 연비 평균 구하기
step4 = step3.agg(mean_total = ('total', 'mean'))

# 5. 내림차순 정렬하기
step5 = step4.sort_values('mean_total', ascending = False)

# 6. 1~5위까지 출력하기
step6 = step5.head()

print(step6)
```

```python
# 체이닝을 사용하여 작성한 코드

m2 = mpg.query('category == "suv"') \
        .assign(total = lambda x: x['cty'] + x['hwy']) \
        .groupby('manufacturer') \
        .agg(mean_total = ('total', 'mean')) \
        .sort_values('mean_total', ascending = False) \
        .head()
print(m2)

print("\n단계별 코드와 결과가 같은가?", m2.equals(step6))
```

- **체이닝 안에서는 반드시 lambda 를 사용할 것**
  - `query( )` 로 걸러낸 뒤에는 데이터의 행 수가 달라져 있음
  - 그 상태에서 원본 이름(`mpg`)을 다시 쓰면 **행 수가 맞지 않는 값**을 넣게 됨
  - lambda 의 `x` 는 **바로 앞 단계까지 처리된 결과**를 가리키므로 항상 안전함

- 만약 lambda 를 쓰지 않고 `assign(total = mpg['cty'] + mpg['hwy'])` 로 작성하면?
  - 지금 데이터에서는 인덱스가 그대로 남아 있어 **우연히 올바른 값**이 나옴
  - 그러나 중간에 `reset_index( )` 등이 끼면 **오류 없이 값만 틀리게** 계산됨
- 아래 셀에서 그 차이를 확인할 것

```python
small = pd.DataFrame({'category': ['suv', 'compact', 'suv', 'suv'],
                      'cty': [10, 20, 12, 14],
                      'hwy': [15, 30, 18, 20]})

# 원본 이름을 그대로 쓴 경우 - reset_index 가 끼면 값이 틀어짐
wrong = small.query('category == "suv"') \
             .reset_index(drop = True) \
             .assign(total = small['cty'] + small['hwy'])

# lambda 를 쓴 경우 - 항상 올바름
right = small.query('category == "suv"') \
             .reset_index(drop = True) \
             .assign(total = lambda x: x['cty'] + x['hwy'])

print("원본 이름 사용:")
print(wrong)
print()
print("lambda 사용:")
print(right)
```

### 데이터 합치기
- 여러 데이터를 합쳐 하나로 만든 다음 분석할 필요가 있는 경우
   - 예: 중간고사 데이터와 기말고사 데이터를 합쳐 하나로 만들어 분석

- 가로로 합치기
  - 한 데이터에 변수(열)을 추가하는 형태
  - 예: 중간고사 데이터에 기말고사 데이터를 합치는 경우
  - 기준이 되는 변수(예: 학번)의 값을 맞추어 합칠 때는 `merge( )` 사용

- 세로로 합치기
  - 데이터에 행을 추가하는 형태
  - 예: 학생 3명이 먼저 시험을 보고, 나중에 다른 학생 3명이 시험을 봤을 때 두 시험 데이터를 합치는 경우
  - `concat( )` 사용

- `concat( )` 은 `axis = 1` 을 지정하면 가로로도 합칠 수 있음
  - 다만 기준 변수의 값을 맞추지 않고 **순서대로** 붙이므로, 두 데이터의 행 순서가 대응할 때만 사용


- 가로로 합치기 예제

```python
# 중간고사 데이터 만들기
test1 = pd.DataFrame({'id'      : [1, 2, 3, 4, 5],
                      'midterm' : [60, 80, 70, 90, 85]})

# 기말고사 데이터 만들기
test2 = pd.DataFrame({'id'    : [1, 2, 3, 4, 5],
                      'final' : [70, 83, 65, 95, 80]})

print(test1)
print()
print(test2)
```

```python
total = pd.merge(test1, test2, how = 'left', on = 'id') # id 기준으로 test1과 test2를 합쳐서 total에 할당
print(total)
```

- 세로로 합치기 예제

```python
# 학생 1~5번 시험 데이터 만들기
group_a = pd.DataFrame({'id'   : [1, 2, 3, 4, 5],
                        'test' : [60, 80, 70, 90, 85]})

# 학생 6~10번 시험 데이터 만들기
group_b = pd.DataFrame({'id'   : [6, 7, 8, 9, 10],
                        'test' : [70, 83, 65, 95, 80]})

print(group_a)
print()
print(group_b)
```

### 아래 셀을 실행하기 전에 생각해 볼 것

- `group_a` 와 `group_b` 는 각각 5행이며, 인덱스는 둘 다 `0 ~ 4` 임
- 두 데이터를 세로로 이어 붙이면 **결과의 인덱스는 어떻게 될까?**
  1. 0 ~ 9 로 새로 매겨짐
  2. 0 ~ 4 가 두 번 반복됨
  3. 오류 발생

- 4주차에서 다룬 "인덱스는 행의 이름"이라는 점을 떠올려 볼 것

> 나의 예측:


```python
group_all = pd.concat([group_a, group_b])
print(group_all)
```

```python
# (설명 가리기용 여백 - 실행하지 않아도 됨)











































```

- 각 행은 자신의 인덱스(이름)를 **그대로 가진 채** 결합됨
  - `concat( )` 은 데이터를 이어 붙일 뿐 인덱스를 새로 매기지 않음
  - 그래서 `group_b` 의 0~4 가 그대로 따라와 인덱스가 중복됨
- 순서대로 번호를 새로 매기려면 `ignore_index = True` 를 지정

```python
group_all = pd.concat([group_a, group_b], ignore_index = True) # 인덱스 번호를 새로 부여하기 위해 ignore_index = True 추가
print(group_all)
```

- `axis = 1` 을 지정하면 **가로로** 이어 붙임
  - `merge( )` 는 기준 변수의 **값을 맞추어** 합치지만, `concat(axis=1)` 은 **순서대로** 나란히 붙임
  - 두 데이터의 행 순서가 대응한다는 보장이 없으면 잘못된 결과가 나오므로 주의

```python
side = pd.concat([group_a, group_b], axis = 1)
print(side)

# 같은 열 이름(id, test)이 두 번 나타나는 것에 주의
```

#### merge( ), concat( ) 함수 참고

- `pandas.merge(left, right, how='inner', on=None)`
  - `left`, `right`: 결합할 DataFrame
  - `on`: 병합 기준 변수. 이 변수의 **값이 같은 행끼리** 짝을 지어 합침
    - 이렇게 짝을 맞추는 기준이 되는 값을 키(key) 라고 부름
    - 양쪽에 같은 이름으로 있어야 함
  - `how`: 어느 쪽의 **행**을 남길지 결정

  | how | 남는 행 | 없는 값 |
  |---|---|---|
  | `'inner'` (기본값) | 양쪽에 모두 있는 기준 변수 값 | — |
  | `'outer'` | 양쪽의 모든 기준 변수 값 | `NaN` |
  | `'left'` | 왼쪽의 모든 기준 변수 값 | 오른쪽 열이 `NaN` |
  | `'right'` | 오른쪽의 모든 기준 변수 값 | 왼쪽 열이 `NaN` |


- `pandas.concat(df_list, axis=0, join='outer', ignore_index=False)`
  - `df_list`: 결합할 DataFrame의 리스트
  - `axis`: `0` 이면 행을 늘려서(세로로) 결합, `1` 이면 열을 늘려서(가로로) 결합
  - `join`: 두 데이터의 열 이름이 서로 다를 때의 처리 방식
    - `'outer'` (기본값): 열 이름의 합집합. 없는 칸은 `NaN`
    - `'inner'`: 양쪽에 모두 있는 열만 남김
  - `ignore_index`: `True` 이면 기존 인덱스를 버리고 0부터 새로 매김

- `merge()`로 데이터를 합칠 때 두 데이터의 `id` 값이 **일부만 겹치는 경우**

```python
# 중간고사 데이터 만들기
test1 = pd.DataFrame({'id'      : [1, 2, 3, 4, 5],
                      'midterm' : [60, 80, 70, 90, 85]})

# 기말고사 데이터 만들기
test2 = pd.DataFrame({'id'    : [3, 4, 5, 6, 7],
                      'final' : [70, 83, 65, 95, 80]})
```

#### 아래 셀들을 실행하기 전에 생각해 볼 것

- `test1` 의 id 는 1, 2, 3, 4, 5 이고 `test2` 의 id 는 3, 4, 5, 6, 7 임
- 겹치는 id 는 3, 4, 5 뿐임
- `how` 값에 따라 결과의 **행 수**가 어떻게 달라질지 먼저 예측해 볼 것
- `NaN` 이 생긴다면 **어느 쪽 열**에 생길지도 함께 생각할 것

| how | 예상 행 수 | 남는 id |
|---|---|---|
| left | | |
| right | | |
| inner | | |
| outer | | |

> 나의 예측:


```python
# id 기준으로 test1과 test2를 합쳐서 total에 할당
# how = 'left' : 왼쪽 DataFrame의 모든 키를 유지. 오른쪽에 없는 값은 NaN으로 채움.
total = pd.merge(test1, test2, how = 'left', on = 'id') 
print(total)
```

```python
# id 기준으로 test1과 test2를 합쳐서 total에 할당
# how = 'right' : 오른쪽 DataFrame의 모든 키를 유지. 왼쪽에 없는 값은 NaN으로 채움.
total = pd.merge(test1, test2, how = 'right', on = 'id')
print(total)
```

```python
# id 기준으로 test1과 test2를 합쳐서 total에 할당
# how = 'inner' : 두 DataFrame에서 겹치는 키 값만 결과에 포함. 공통 부분만 보고 싶을 때 사용.
total = pd.merge(test1, test2, how = 'inner', on = 'id') 
print(total)
```

```python
# id 기준으로 test1과 test2를 합쳐서 total에 할당
# how = 'outer' : 양쪽 DataFrame의 모든 키를 포함. 없는 부분은 NaN으로 채움.
total = pd.merge(test1, test2, how = 'outer', on = 'id') 
print(total)
```

- `concat()`으로 세로로 합칠 때 **열 이름이 일부만 겹치는 경우**
  - 1차 시험에는 점수만 기록했는데, 2차 시험부터는 가산점(`bonus`)도 함께 기록한 상황
  - `join` 값에 따라 처리 방식이 달라짐

```python
# 1차 시험 - 점수만 기록
group_a = pd.DataFrame({'id'   : [1, 2, 3],
                        'test' : [60, 80, 70]})

# 2차 시험 - 가산점 항목이 추가됨
group_c = pd.DataFrame({'id'    : [4, 5, 6],
                        'test'  : [90, 85, 75],
                        'bonus' : [5, 0, 10]})

print(group_a)
print()
print(group_c)
```

```python
# join = 'outer' (기본값) - 열 이름의 합집합. 없는 칸은 NaN
out = pd.concat([group_a, group_c], ignore_index = True)
print(out)
print()
print(out.dtypes)
```

```python
# join = 'inner' - 양쪽에 모두 있는 열만 남김
inn = pd.concat([group_a, group_c], join = 'inner', ignore_index = True)
print(inn)
```

- `'outer'` 는 정보를 잃지 않지만 `NaN` 이 생김
- `'inner'` 는 `NaN` 이 없는 대신 `bonus` 열이 통째로 사라짐
- 어느 쪽이 맞는지는 분석 목적에 따라 판단해야 함
